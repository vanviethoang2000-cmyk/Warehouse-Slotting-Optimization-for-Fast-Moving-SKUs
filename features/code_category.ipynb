```python
import pandas as pd
import numpy as np
import os
current_dir = os.getcwd()

# 2. Dùng os.path.join để tự động nối tên thư mục với tên file data
# Máy Window sẽ tự nối bằng dấu '\', máy Mac/Linux sẽ tự nối bằng dấu '/'
file_path = os.path.join(current_dir, 'data_clean.csv')

print(f"Đường dẫn file máy tự tìm được là: {file_path}")

# 3. Đọc dữ liệu
df_clean = pd.read_csv(file_path, sep=';')

def assign_category(text):
    if pd.isna(text):
        return 'Others'
    
    text = text.upper()
    
    # Bỏ chữ 'GLASS' chung chung ra khỏi nhóm bếp
    kitchen_keywords = ['MUG', 'CUP', 'BOWL', 'PLATE', 'CAKESTAND', 'TIN', 
                        'CAKE', 'BAKING', 'PAN', 'FORK', 'SPOON', 'KNIFE', 
                        'CUTLERY', 'APRON', 'JAR']
                        
    lighting_keywords = ['LIGHT', 'LAMP', 'LED', 'LANTERN', 'BULB', 'CANDLE']
    bag_keywords = ['BAG', 'SHOPPER', 'TOTE']
    party_keywords = ['HEART', 'ORNAMENT', 'BUNTING', 'FRAME', 'VINTAGE', 'RETRO', 'CHRISTMAS', 'PARTY']
    stationery_keywords = ['WRAP', 'PAPER', 'TAPE', 'PEN', 'RIBBON', 'CARD', 'PENCIL']
    
    # --- THỨ TỰ ƯU TIÊN KIỂM TRA TỪ TRÊN XUỐNG ---
    
    # Ưu tiên 1: Bắt Đèn trước (GLASS LIGHT sẽ bị tóm gọn ở đây)
    if any(word in text for word in lighting_keywords):
        return 'Lighting & Candles'
        
    # Ưu tiên 2: Bắt Túi xách
    elif any(word in text for word in bag_keywords):
        return 'Bags'
        
    # Ưu tiên 3: Bắt Đồ dùng bếp (Thêm cụm từ rõ nghĩa cho thủy tinh)
    elif any(word in text for word in kitchen_keywords) or 'WINE GLASS' in text or 'WATER GLASS' in text:
        return 'Tableware & Kitchen'
        
    # Ưu tiên 4: Trang trí tiệc
    elif any(word in text for word in party_keywords):
        return 'Party & Decoration'
        
    # Ưu tiên 5: Văn phòng phẩm
    elif any(word in text for word in stationery_keywords):
        return 'Stationery & Crafts'
        
    # Chốt sổ
    else:
        return 'Others'
df_clean['category'] = df_clean['description'].apply(assign_category)
print(df_clean[['description','category']].head(100))
print(df_clean['category'].value_counts())
```

```python
# Lưu bảng dữ liệu đã làm sạch và có thêm cột category ra file CSV
df_clean.to_csv('cleaned_online_retail.csv', index=False)

print("Đã xuất file thành công!")
```
