```python
import os

# =====================================================================
# 1. DANH SÁCH THƯ MỤC CẦN TẠO (CHUẨN COOKIECUTTER DATA SCIENCE)
# =====================================================================
folders = [
    'data/raw',                 # Chứa file raw_data.xlsx gốc
    'data/processed',           # Chứa train_invoices, test_invoices
    'features',                 # Chứa sku_features.csv, sku_clusters_final.csv
    'notebooks',                # Chứa các file .ipynb (01, 02, 03)
    'src/utils',                # Chứa các hàm code phụ trợ
    'src/models',               # Chứa slotting_algorithms.py
    'src/envs',                 # Chứa warehouse_env.py (Môi trường ảo)
    'sql',                      # Chứa các file script .sql
    'reports/figures',          # Nơi lưu ảnh biểu đồ (Heatmap, Boxplot...) xuất ra từ code
    'reports/forecasts',        # Kết quả dự báo/mô phỏng
    'dashboard',                # Chứa file Power BI (.pbix)
    'docs',                     # Tài liệu đồ án (Word, PDF báo cáo)
]

# =====================================================================
# 2. DANH SÁCH CÁC FILE TRỐNG CẦN TẠO
# =====================================================================
files = [
    '.gitignore', 
    'README.md', 
    'requirements.txt', 
    'src/__init__.py',
    'src/utils/__init__.py',
    'src/models/__init__.py',
    'src/envs/__init__.py'
]

# =====================================================================
# 3. THỰC THI TẠO CẤU TRÚC
# =====================================================================
print("Đang khởi tạo cấu trúc dự án chuẩn Software Engineering...")

# Tạo thư mục
for folder in folders:
    os.makedirs(folder, exist_ok=True)
    print(f"📁 Đã tạo thư mục: {folder}")

# Tạo file
for file in files:
    # Nếu là file .gitignore, ghi sẵn vài dòng cấu hình cơ bản để không đẩy rác lên Github
    if file == '.gitignore':
        with open(file, 'w', encoding='utf-8') as f:
            f.write("__pycache__/\n*.ipynb_checkpoints/\ndata/raw/\n.env\n")
        print(f"📄 Đã tạo file: {file} (Kèm cấu hình cơ bản)")
    else:
        with open(file, 'w', encoding='utf-8') as f:
            pass 
        print(f"📄 Đã tạo file: {file}")

print("\n✅ [THÀNH CÔNG] Toàn bộ thư mục và file dự án đã được tạo hoàn tất!")
```
